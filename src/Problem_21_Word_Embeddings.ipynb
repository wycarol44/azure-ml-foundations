```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

sentences = [
    "I love machine learning and artificial intelligence.",
    "I enjoy studying AI and machine learning.",
    "The weather is sunny and warm today.",
    "Deep learning is a branch of artificial intelligence."
]

# 1. TF-IDF representation
tfidf = TfidfVectorizer()
tfidf_vectors = tfidf.fit_transform(sentences)

tfidf_similarity = cosine_similarity(tfidf_vectors)

print("TF-IDF Similarity:")
print(tfidf_similarity.round(3))

# 2. Dense sentence embeddings
model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(sentences)

embedding_similarity = cosine_similarity(embeddings)

print("\nSentence Embedding Similarity:")
print(embedding_similarity.round(3))
```
